```python
#Setup
import pandas as pd
marks = pd.read_csv("/kaggle/input/datasets/rebalmorshed/marks1/StudentsPerformance.csv", index_col=0)
from learntools.core import binder; binder.bind(globals())
print(marks.head())
```

```python
#Understanding the data
print (marks.info())
print (marks.describe())
```

```python
#Checking for null values
print (marks.isnull().sum())
```

***We can see that there is no null values***

```python
#Print the average of math,reading and writing scores
print(marks[['math score','reading score','writing score']].mean())
```

***Reading scores are generally higher than math scores***

```python
#Print math scores that are upov 80
high_math = marks[marks['math score']>80]
print(high_math)
```

***We type ( .... = marks[marks[,,,,,]] ) to print just the true values(marks upov 80), and avoid printing (true / false) for all the 1000 students***

```python
#Visualizatoin
import matplotlib.pyplot as plt
marks['math score'].hist()
plt.title('Math Score Distribution')
plt.show()
```

***We can see that there are about 250 students who recieved scores between 60 and 80 in math***

```python
#Test preparation course for math
marks.groupby('test preparation course')['math score'].mean()
```

***Completed courses is more than the not completed***

```python
#comparing male and female scores in math
marks.groupby('gender')['math score'].mean()
```

***Males average marks in math is higher than females marks***

```python
#Visualization 2
import matplotlib.pyplot as plt2
marks.groupby('gender')['math score'].mean().plot(kind='bar')
plt2.title('Average Math Score by Gender')
plt2.show()
```

***We usded a more advanced way to visualize the comparison between males and females marks in math***
